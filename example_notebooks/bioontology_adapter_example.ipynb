# BioOntologyAdapter Example
Demonstrates basic usage of the BioOntologyAdapter.

```python
from aid_pais_knowledgegraph.knowledge_lookup.adapters.bioontology_adapter import BioOntologyAdapter
from aid_pais_knowledgegraph.knowledge_lookup.models import LookupConfig


# Initialize adapter with a config
# You can pass the API key directly, or leave api_keys empty to use .env fallback.
config = LookupConfig()  # .env will be used if available

# Use async context manager for proper cleanup
async with BioOntologyAdapter(config) as adapter:
    # Always pass the search query as the first argument
    results = await adapter.search_concepts(query='Chronic Fatigue Syndrome', limit=15)
for concept in results:
    print(concept.primary_id, concept.primary_label)
    print(concept)
```

```python
async with BioOntologyAdapter(config) as adapter:
    raw_results = await adapter.search_concepts('Chronic Fatigue Syndrome', limit=15, raw=True)
```

```python
# Convert raw_results to RDF using rdflib
from rdflib import Graph, URIRef, Literal, Namespace
from rdflib.namespace import RDF, RDFS, SKOS, OWL, DC, DCTERMS, FOAF, XSD
from urllib.parse import quote
g = Graph()
EX = Namespace('http://aid-pais.org/')
BIOPORTAL = Namespace('http://data.bioontology.org/ontologies/')
MESH = Namespace('http://id.nlm.nih.gov/mesh/')
OBO = Namespace('http://purl.obolibrary.org/obo/')
g.bind('ex', EX)
g.bind('skos', SKOS)
g.bind('owl', OWL)
g.bind('rdfs', RDFS)
g.bind('dc', DC)
g.bind('dcterms', DCTERMS)
g.bind('foaf', FOAF)
g.bind('xsd', XSD)
g.bind('bioportal', BIOPORTAL)
g.bind('mesh', MESH)
g.bind('obo', OBO)
for item in raw_results:
    # Subject URI
    if '@id' in item:
        subj = URIRef(item['@id'])
    else:
        label = item.get('prefLabel', '')
        safe_label = quote(label.replace(' ', '_'))
        subj = URIRef(EX[f'concept_{safe_label}'])
    # Type
    g.add((subj, RDF.type, URIRef(item.get('@type', OWL.Class))))
    # PrefLabel
    if 'prefLabel' in item:
        g.add((subj, SKOS.prefLabel, Literal(item['prefLabel'])))
    # Synonyms
    if 'synonym' in item:
        for s in item['synonym'] if isinstance(item['synonym'], list) else [item['synonym']]:
            g.add((subj, SKOS.altLabel, Literal(s)))
    # Definition
    if 'definition' in item:
        for d in item['definition'] if isinstance(item['definition'], list) else [item['definition']]:
            g.add((subj, SKOS.definition, Literal(d)))
    # MatchType
    if 'matchType' in item:
        g.add((subj, EX.matchType, Literal(item['matchType'])))
    # SemanticType
    if 'semanticType' in item:
        for st in item['semanticType'] if isinstance(item['semanticType'], list) else [item['semanticType']]:
            g.add((subj, EX.semanticType, Literal(st)))
    # CUI
    if 'cui' in item:
        for c in item['cui'] if isinstance(item['cui'], list) else [item['cui']]:
            g.add((subj, EX.cui, Literal(c)))
    # Links (parents, children, ancestors, descendants, ontology, mappings, ui, notes)
    links = item.get('links', {})
    for rel, url in links.items():
        if rel in ['parents', 'children', 'ancestors', 'descendants', 'ontology', 'mappings', 'ui', 'notes']:
            pred = EX[rel]
            g.add((subj, pred, URIRef(url)))
# Show as Turtle
print(g.serialize(format='turtle').decode() if hasattr(g.serialize(format='turtle'), 'decode') else g.serialize(format='turtle'))
```

```python
# Export RDF graph as Turtle (.ttl) file
with open('bioontology_results.ttl', 'w') as f:
    f.write(g.serialize(format='turtle'))
print('RDF graph exported to bioontology_results.ttl')
```

```python
# Fetch and display related resources (children, ancestors, etc.) for a concept, with URL debugging
from pprint import pprint
async with BioOntologyAdapter(config) as adapter:
    if results:
        concept = results[0]
        print(concept)
        concept_id = concept.primary_id
        ontology = concept_id.split('/')[-2]
        print(ontology)
        # Fetch details with related resources
        detailed = await adapter.get_concept_details(concept_id, ontology=ontology, fetch_related=True, minimal=True)
        if detailed is not None:
            # Print related resources and URLs before API call
            for key in ['bioontology_children', 'bioontology_ancestors', 'bioontology_parents', 'bioontology_descendants']:
                url_key = key.replace('bioontology_', '')
                if hasattr(detailed, 'source_data') and 'BIOONTOLOGY' in detailed.source_data:
                    links = detailed.source_data['BIOONTOLOGY'].get('links', {})
                    if url_key in links:
                        print(f'URL for {key}: {links[url_key]}')
                if key in detailed.source_data:
                    print(f'--- {key} ---')
                    pprint(detailed.source_data[key])
        else:
            print('No details found for concept.')
    else:
        print('No concepts found in results.')
```

```python
# Example: Safely handle both dict and object returns from get_concept_details
def print_related_resources(detailed):
    from pprint import pprint
    # Try to get source_data if present (parsed object), else use dict directly
    source_data = getattr(detailed, 'source_data', None) if hasattr(detailed, 'source_data') else None
    if source_data and 'BIOONTOLOGY' in source_data:
        links = source_data['BIOONTOLOGY'].get('links', {})
    elif isinstance(detailed, dict) and 'links' in detailed:
        links = detailed.get('links', {})
    else:
        links = {}
    for key in ['bioontology_children', 'bioontology_ancestors', 'bioontology_parents', 'bioontology_descendants']:
        url_key = key.replace('bioontology_', '')
        if url_key in links:
            print(f'URL for {key}: {links[url_key]}')
        # Print related data if present (for parsed object or dict)
        if source_data and key in source_data:
            print(f'--- {key} ---')
            pprint(source_data[key])
        elif isinstance(detailed, dict) and key in detailed:
            print(f'--- {key} ---')
            pprint(detailed[key])
print('Related resources:')
print_related_resources(detailed)
```
