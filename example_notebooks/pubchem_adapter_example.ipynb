# PubChem Adapter Example
Demonstrates searching and retrieving compound details using the PubChemAdapter.

```python
from aid_pais_knowledgegraph.knowledge_lookup.adapters.pubchem_adapter import PubChemAdapter
import asyncio
adapter = PubChemAdapter()
results = asyncio.run(adapter.search_concepts('aspirin', limit=5))
print('Compound IDs:', results)
```

```python
cid = results[0] if results else None
if cid:
    details = asyncio.run(adapter.get_concept_details(cid))
    print('Details:', details)
```
