# MONDOAdapter Example
Demonstrates basic usage of the MONDOAdapter.

```python
from aid_pais_knowledgegraph.knowledge_lookup.adapters.mondo_adapter import MONDOAdapter
from aid_pais_knowledgegraph.knowledge_lookup.models import LookupConfig

# Initialize adapter with a config
config = LookupConfig()
adapter = MONDOAdapter(config)

# Search for concepts
import asyncio
results = asyncio.run(adapter.search_concepts('fatigue', limit=5))
for concept in results:
    print(concept.primary_id, concept.primary_label)
```
