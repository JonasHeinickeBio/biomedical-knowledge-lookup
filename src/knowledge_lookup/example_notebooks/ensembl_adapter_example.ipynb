# EnsemblAdapter Example
Demonstrates basic usage of the EnsemblAdapter.

```python
from aid_pais_knowledgegraph.knowledge_lookup.adapters.ensembl_adapter import EnsemblAdapter
from aid_pais_knowledgegraph.knowledge_lookup.models import LookupConfig

# Initialize adapter with a config
config = LookupConfig()
adapter = EnsemblAdapter(config)

# Search for concepts
results = await adapter.search_concepts('fatigue', limit=5)
for concept in results:
    print(concept.primary_id, concept.primary_label)
```
