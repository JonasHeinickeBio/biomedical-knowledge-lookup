## DisGeNET Score Explanation
The DisGeNET gene-disease association results include several scores that help interpret the relevance and specificity of each gene to the disease:
- **gene_dsi (Disease Specificity Index):** Measures how specific a gene is to a particular disease. A higher value means the gene is associated with fewer diseases, indicating higher specificity.
- **gene_dpi (Disease Pleiotropy Index):** Indicates how many different diseases a gene is associated with. A higher value means the gene is linked to more diseases, reflecting pleiotropy.
- **gene_pli (Gene Pleiotropy Index):** Represents the overall pleiotropy of the gene across all its disease associations. A value close to 1 means the gene is highly pleiotropic, associated with many diseases.
These scores help prioritize gene-disease associations for further study, with high specificity (DSI) suggesting disease relevance and high pleiotropy (DPI/PLI) indicating broader biological roles.

```python
from rdflib import Graph, URIRef, Literal, RDF, RDFS, Namespace, OWL, XSD
from datetime import datetime

# Define namespaces
AIDPAIS = Namespace("http://aid-pais.org/ontology#")
RO = Namespace("http://purl.obolibrary.org/obo/RO_")
SO = Namespace("http://purl.obolibrary.org/obo/SO_")
UNIPROT = Namespace("https://identifiers.org/uniprot:")
NCBIGENE = Namespace("https://identifiers.org/ncbigene:")
UMLS = Namespace("https://identifiers.org/umls:")
DISGENET = Namespace("https://disgenet.org/association/")
BIOLINK = Namespace("https://w3id.org/biolink/vocab/")
GO = Namespace("http://purl.obolibrary.org/obo/GO_")
MESH = Namespace("https://identifiers.org/mesh:")
ICD10 = Namespace("https://identifiers.org/icd10:")
NCI = Namespace("https://identifiers.org/ncit:")
ICD9CM = Namespace("https://identifiers.org/icd9cm:")
DOID = Namespace("https://identifiers.org/doid:")
MONDO = Namespace("https://identifiers.org/mondo:")
EFO = Namespace("https://identifiers.org/efo:")
PROV = Namespace("http://www.w3.org/ns/prov#")

# Create graph and bind namespaces
g = Graph()
g.bind("aidpais", AIDPAIS)
g.bind("ro", RO)
g.bind("so", SO)
g.bind("uniprot", UNIPROT)
g.bind("ncbigene", NCBIGENE)
g.bind("umls", UMLS)
g.bind("disgenet", DISGENET)
g.bind("biolink", BIOLINK)
g.bind("go", GO)
g.bind("mesh", MESH)
g.bind("icd10", ICD10)
g.bind("ncit", NCI)
g.bind("icd9cm", ICD9CM)
g.bind("doid", DOID)
g.bind("mondo", MONDO)
g.bind("efo", EFO)
g.bind("rdfs", RDFS)
g.bind("owl", OWL)
g.bind("prov", PROV)

# Create a provenance node for this query and service
provenance_uri = URIRef("http://aid-pais.org/provenance/disgenet_query_" + datetime.now().strftime("%Y%m%d%H%M%S"))
g.add((provenance_uri, RDF.type, PROV.Activity))
g.add((provenance_uri, RDFS.comment, Literal("Provenance for DisGeNETAdapter query.")))
g.add((provenance_uri, RDFS.label, Literal("DisGeNETAdapter Query Provenance")))
g.add((provenance_uri, AIDPAIS.service, Literal("aid_pais_knowledgegraph.knowledge_lookup.adapters.disgenet_adapter.DisGeNETAdapter")))
g.add((provenance_uri, AIDPAIS.query, Literal(str(params))))
g.add((provenance_uri, AIDPAIS.time, Literal(datetime.now().isoformat(), datatype=XSD.dateTime)))
g.add((provenance_uri, AIDPAIS.score_explanation, Literal("gene_dsi (Disease Specificity Index): Measures how specific a gene is to a particular disease. Higher values mean greater specificity.")))
g.add((provenance_uri, AIDPAIS.score_explanation, Literal("gene_dpi (Disease Pleiotropy Index): Indicates how many diseases a gene is associated with. Higher values mean more associations.")))
g.add((provenance_uri, AIDPAIS.score_explanation, Literal("gene_pli (Gene Pleiotropy Index): Represents overall pleiotropy of the gene across all disease associations. Values near 1 indicate high pleiotropy.")))

for assoc in results:
    gene_uri = NCBIGENE[str(assoc['gene_ncbi_id'])]
    disease_uri = UMLS[str(assoc['disease_umls_cui'])]
    assoc_uri = DISGENET[str(assoc['assocID'])]

    # Use AID-PAIS classes and properties
    g.add((gene_uri, RDF.type, AIDPAIS.Gene))
    g.add((disease_uri, RDF.type, AIDPAIS.Disease))
    g.add((assoc_uri, RDF.type, AIDPAIS.GeneDiseaseAssociation))
    g.add((assoc_uri, AIDPAIS.participatesIn, disease_uri))
    g.add((assoc_uri, AIDPAIS.hasParticipant, gene_uri))
    g.add((assoc_uri, AIDPAIS.score, Literal(assoc['score'])))
    g.add((assoc_uri, AIDPAIS.num_pmids, Literal(assoc['num_pmids'])))

    # Add DisGeNET scores and link to provenance node
    g.add((assoc_uri, AIDPAIS.gene_dsi, Literal(assoc.get('gene_dsi', None))))
    g.add((assoc_uri, AIDPAIS.gene_dpi, Literal(assoc.get('gene_dpi', None))))
    g.add((assoc_uri, AIDPAIS.gene_pli, Literal(assoc.get('gene_pli', None))))
    g.add((assoc_uri, PROV.wasDerivedFrom, provenance_uri))

    # Add gene_ensembl_ids and gene_type to the KG
    for ensembl_id in assoc.get('gene_ensembl_ids', []):
        g.add((gene_uri, AIDPAIS.ensembl_id, Literal(ensembl_id)))
    g.add((gene_uri, AIDPAIS.gene_type, Literal(assoc.get('gene_type', None))))

    # Add protein information coded by the gene
    for protein_id in assoc.get('gene_protein_str_ids', []):
        protein_uri = UNIPROT[str(protein_id)]
        g.add((protein_uri, RDF.type, AIDPAIS.Protein))
        g.add((gene_uri, AIDPAIS.hasGeneProduct, protein_uri))
        # Move protein_class to the protein node
        for protein_class in assoc.get('gene_protein_class_names', []):
            g.add((protein_uri, AIDPAIS.protein_class, Literal(protein_class)))

    # Optionally add labels
    g.add((gene_uri, RDFS.label, Literal(assoc['gene_symbol'])))
    g.add((disease_uri, RDFS.label, Literal(assoc['disease_name'])))

    # Add additional disease IDs to the disease node using owl:sameAs for equivalence
    same_as_uris = []
    for vocab_id in assoc.get('disease_vocabularies', []):
        if vocab_id.startswith('MESH_'):
            mesh_id = vocab_id.replace('MESH_', '')
            same_as_uris.append(MESH[mesh_id])
        elif vocab_id.startswith('ICD10_'):
            icd10_id = vocab_id.replace('ICD10_', '')
            same_as_uris.append(ICD10[icd10_id])
        elif vocab_id.startswith('NCI_'):
            nci_id = vocab_id.replace('NCI_', '')
            same_as_uris.append(NCI[nci_id])
        elif vocab_id.startswith('ICD9CM_'):
            icd9cm_id = vocab_id.replace('ICD9CM_', '')
            same_as_uris.append(ICD9CM[icd9cm_id])
        elif vocab_id.startswith('DO_'):
            doid_id = vocab_id.replace('DO_', '')
            same_as_uris.append(DOID[doid_id])
        elif vocab_id.startswith('MONDO_'):
            mondo_id = vocab_id.replace('MONDO_', '')
            same_as_uris.append(MONDO[mondo_id])
        elif vocab_id.startswith('EFO_'):
            efo_id = vocab_id.replace('EFO_', '')
            same_as_uris.append(EFO[efo_id])
        elif vocab_id.startswith('UMLS_'):
            umls_id = vocab_id.replace('UMLS_', '')
            same_as_uris.append(UMLS[umls_id])

    # Add owl:sameAs triples for all equivalent disease URIs
    for uri in same_as_uris:
        g.add((disease_uri, OWL.sameAs, uri))

print(g.serialize(format='turtle'))
```

```python
# Export RDF graph as Turtle (.ttl) file
with open('disgenet_results.ttl', 'w') as f:
    f.write(g.serialize(format='turtle'))
print('RDF graph exported to disgenet_results.ttl')
```

```python
# Enrich KG using selected UniProt columns for each protein
from rdflib import URIRef, Literal, RDF, RDFS, OWL, Namespace
import pandas as pd

# Namespaces (reuse from above)
UNIPROT = Namespace("https://identifiers.org/uniprot:")
SKOS = Namespace("http://www.w3.org/2004/02/skos/core#")

for idx, row in protein_df.iterrows():
    protein_uri = UNIPROT[row['Entry']]
    # Core identifiers
    g.add((protein_uri, RDF.type, URIRef("http://aid-pais.org/ontology#Protein")))
    g.add((protein_uri, RDFS.label, Literal(row.get('Entry Name', row['Entry']))))
    # Gene and protein names
    # Find gene by hasGeneProduct relationship
    for gene_uri in g.subjects(URIRef("http://aid-pais.org/ontology#hasGeneProduct"), protein_uri):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#isGeneProductOf"), gene_uri))
    if pd.notnull(row.get('Gene Names')):
        for gene in str(row['Gene Names']).split(';'):
            g.add((protein_uri, SKOS.altLabel, Literal(gene.strip())))
    if pd.notnull(row.get('Protein names')):
        for pname in str(row['Protein names']).split(';'):
            g.add((protein_uri, SKOS.altLabel, Literal(pname.strip())))
    # Proteomes
    if pd.notnull(row.get('Proteomes')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#proteomes"), Literal(row['Proteomes'])))
    # Length and mass
    if pd.notnull(row.get('Length')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#sequenceLength"), Literal(row['Length'])))
    if pd.notnull(row.get('Mass')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#molecularWeight"), Literal(row['Mass'])))
    # Protein families
    if pd.notnull(row.get('Protein families')):
        for fam in str(row['Protein families']).split(';'):
            g.add((protein_uri, URIRef("http://aid-pais.org/ontology#protein_family"), Literal(fam.strip())))
    # Binding site
    if pd.notnull(row.get('Binding site')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#binding_site"), Literal(row['Binding site'])))
    # Catalytic activity
    if pd.notnull(row.get('Catalytic activity')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#catalytic_activity"), Literal(row['Catalytic activity'])))
    # Cofactor
    if pd.notnull(row.get('Cofactor')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#cofactor"), Literal(row['Cofactor'])))
    # DNA binding
    if pd.notnull(row.get('DNA binding')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#dna_binding"), Literal(row['DNA binding'])))
    # Function [CC]
    if pd.notnull(row.get('Function [CC]')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#function"), Literal(row['Function [CC]'])))
    # Pathway
    if pd.notnull(row.get('Pathway')):
        g.add((protein_uri, URIRef("http://aid-pais.org/ontology#pathway"), Literal(row['Pathway'])))
    # Interacts with
    if pd.notnull(row.get('Interacts with')):
        for interactor in str(row['Interacts with']).split(';'):
            interactor_id = interactor.strip()
            if interactor_id:
                interactor_uri = UNIPROT[interactor_id]
                g.add((protein_uri, URIRef("http://aid-pais.org/ontology#interacts_with"), interactor_uri))

print('Selected UniProt columns enrichment added to KG.')
print(g.serialize(format='turtle'))
```

```python
# Export RDF graph as Turtle (.ttl) file
with open('disgenet_with_uniprot.ttl', 'w') as f:
    f.write(g.serialize(format='turtle'))
print('RDF graph exported to disgenet_results.ttl')
```
