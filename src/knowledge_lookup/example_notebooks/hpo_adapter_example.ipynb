# HPOAdapter Example
Demonstrates basic usage of the HPOAdapter.

```python
from aid_pais_knowledgegraph.knowledge_lookup.adapters.hpo_adapter import HPOAdapter
from aid_pais_knowledgegraph.knowledge_lookup.models import LookupConfig

# Initialize adapter with a config
config = LookupConfig()
adapter = HPOAdapter(config)

# Search for concepts
import asyncio
results = asyncio.run(adapter.search_concepts('fatigue', limit=5))
for concept in results:
    print(concept.primary_id, concept.primary_label)
```
